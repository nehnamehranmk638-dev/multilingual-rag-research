```python
from datasets import load_dataset

squad = load_dataset("rajpurkar/squad")
print(squad)
```

```python
val = squad["validation"]

print("Number of validation rows:", len(val))

for i in range(3):
    row = val[i]
    print("QUESTION:", row["question"])
    print("ANSWERS :", row["answers"])
    print("CONTEXT :", row["context"][:200], "...")
    print("-" * 40)
```

```python
print(row.keys())
```

```python

all_contexts = val["context"]

unique_contexts = list(dict.fromkeys(all_contexts))

print("rows:", len(all_contexts))
print("unique passages:", len(unique_contexts))
```

```python
corpus_texts = unique_contexts[:1000]

corpus = [{"passage_id": i, "text": t} for i, t in enumerate(corpus_texts)]

context_to_id = {t: i for i, t in enumerate(corpus_texts)}

print("Number of passages in corpus:", len(corpus))
print(corpus[0])
```

```python
import random

random.seed(42)

candidates = []

for row in val:
    if row["context"] in context_to_id:
        candidates.append({
            "question_id": row["id"],
            "question": row["question"],
            "gold_answers": row["answers"]["text"],
            "gold_passage_id": context_to_id[row["context"]],
        })

print("candidate questions:", len(candidates))

questions = random.sample(candidates, 100)

print("selected questions:", len(questions))
```

```python
for q in questions:
    passage = corpus[q["gold_passage_id"]]["text"]
    assert any(a in passage for a in q["gold_answers"]), q["question_id"]

print("All checks passed")

for q in questions[:3]:
    print(q["question"])
    print("Gold:", q["gold_answers"])
    print("Passage:", corpus[q["gold_passage_id"]]["text"][:200])
    print("-" * 40)
```

```python
import json
import os

os.makedirs("data", exist_ok=True)

with open("data/corpus.json", "w", encoding="utf-8") as f:
    json.dump(corpus, f, ensure_ascii=False, indent=2)

with open("data/questions.json", "w", encoding="utf-8") as f:
    json.dump(questions, f, ensure_ascii=False, indent=2)

print("Files saved successfully!")
```
